**Topics:** Variables, Conditions, Loops, Functions, Lists, Dictionaries

**Cell 1 — Variables & Data Types**

```python
# Variables and basic data types
name = "Alice"
age = 25
height = 5.6
is_student = True

print(name, age, height, is_student)
print(type(name), type(age), type(height))
```

**Cell 2 — Conditions**

```python
# If/elif/else
score = 85

if score >= 90:
    print("Grade: A")
elif score >= 75:
    print("Grade: B")
else:
    print("Grade: C")
```

**Cell 3 — Loops**

```python
# For loop
for i in range(1, 6):
    print(f"Number: {i}")

# While loop
count = 0
while count < 3:
    print(f"Count: {count}")
    count += 1
```

**Cell 4 — Functions**

```python
# Defining and calling functions
def greet(name):
    return f"Hello, {name}!"

def square(n):
    return n ** 2

print(greet("Alice"))
print(square(4))
```

**Cell 5 — Lists**

```python
# List operations
fruits = ["apple", "banana", "cherry"]

fruits.append("mango")
print(fruits)

# Loop through a list
for fruit in fruits:
    print(fruit.upper())

# List comprehension
squares = [x**2 for x in range(1, 6)]
print(squares)
```

**Cell 6 — Dictionaries**

```python
# Dictionary operations
student = {
    "name": "Alice",
    "age": 25,
    "courses": ["Math", "Data Science"]
}

print(student["name"])
student["grade"] = "A"

# Loop through key-value pairs
for key, value in student.items():
    print(f"{key}: {value}")
```
